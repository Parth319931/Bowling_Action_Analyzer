# Exploratory Data Analysis (EDA) for Bowling Action Analyzer

In this notebook, we will perform exploratory data analysis on the bowling action dataset. The goal is to understand the data better, visualize key features, and identify any patterns or anomalies.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set(style='whitegrid')
```

```python
# Load the dataset
data_path = '../data/processed/bowling_data.csv'  # Update with the correct path
df = pd.read_csv(data_path)

# Display the first few rows of the dataset
df.head()
```

```python
# Summary statistics
df.describe()
```

```python
# Check for missing values
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualize the distribution of key features
plt.figure(figsize=(12, 6))
sns.histplot(df['feature_name'], bins=30, kde=True)
plt.title('Distribution of Feature Name')
plt.xlabel('Feature Name')
plt.ylabel('Frequency')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the bowling action dataset. We visualized the distribution of key features and examined the correlation between them. Further analysis can be conducted based on these insights.
